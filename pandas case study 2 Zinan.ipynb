```python
import pandas as pd
```

```python
#1-
scores=pd.Series([78,85,92,67,74,88,95,61],index=['anu','binu','cathy','deepa','ebin','fathima','gopal','hari'])
print(scores)
```

```python
#9-
print(f'the max score is:{scores.max()}')
print(f'the min score is:{scores.min()}')
print(f'the mean score is:{round(scores.mean(),2)}')
print(f'the count is:{scores.count()}')
```

```python
#10-
print(scores[scores>=80])
```

```python
#11-
v=scores.sort_values(ascending=False)
print(v[0:3])
```

```python
#12-
print(f'the highest student is:{scores.idxmax()}')
```

```python
#13-
def grade(scores):
    if scores>=90:
        return('Distinction')
    elif scores>=75 & scores<=89:
        return('First Class')
    elif scores>=60 & scores<=74:
        return('Second Class')
    else:
        return('Fail')
print(scores.apply(grade))
```

```python
#14-
print(f'the students in each grade is:{scores.value_counts()}')
print(f'the percentage of students in each grade is:{(scores.value_counts(normalize=True)*100)}')
```

```python
#15-
print(f'the unique grades are:{scores.unique()}')
print(f'the no of unique grades are:{scores.nunique()}')
```

```python
#16-
bonus_scores=lambda scores:scores+5
print(scores.apply(bonus_scores))
print(scores)
```
