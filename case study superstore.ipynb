```python
import pandas as pd
```

```python
cf=pd.read_csv('Superstore.csv', encoding='latin1')
print(cf)
```

```python
#q1:-what is the size and structure of the dataset?
print(f'size of dataset is:{cf.size}')
print()
print(f'structure of dataset is:{cf.shape}')
print()
print(cf.info)
```

```python
#2:- Are there missing values or duplicate records?
print(cf.isnull().sum())
print()
print(cf.isnull().sum().sum())
print()
print(cf.duplicated())
```

```python
#3:- What are the total sales, total profit, and average order-line sales?
print('total sales:',cf['Sales'].sum())
print()
print('total profit is',cf['Profit'].sum())
print()
print('avg_sales is',cf['Sales'].mean())
```

```python
#4:- Which product categories generate the most sales?
category_sales=cf.groupby('Category')['Sales'].max()
print(category_sales.sort_values(ascending=False))
```

```python
#5.	Which sub-categories generate the highest and lowest profit?
profit_analysis=cf.groupby('Sub-Category')['Profit'].sum()
print(profit_analysis.sort_values(ascending=False))
print()
print(profit_analysis.idxmax())
print()
print(profit_analysis.idxmin())
```

```python
#6.	Which region has the highest sales?

highest_sales=cf.groupby('Region')['Sales'].sum()
print(highest_sales.sort_values(ascending=False))
print()
print(highest_sales.idxmax())
```

```python
#7.	Which customer segment contributes the most sales?
most_sales=cf.groupby('Segment')['Sales'].max()
print(most_sales.sort_values(ascending=False))
print()
print(most_sales.idxmax())
```

```python
#8.	How do discounts relate to profit?
disc_profit=cf.groupby('Discount')['Profit'].max()
print(disc_profit.sort_values(ascending=False))
```
